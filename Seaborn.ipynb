```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names() #these are the in-bulit datasets of seaborn
```

```python
penguins=sns.load_dataset("penguins")
```

```python
penguins
```

```python
penguins["species"].value_counts()
```

```python
penguins.isnull().sum()
```

```python
penguins["island"].value_counts()
```

```python
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g")
```

```python
sns.set_style("ticks")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("whitegrid")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("darkgrid")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("white")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("dark")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
```

```python
sns.set_style("ticks")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```

```python
sns.set_style("ticks")
sns.set_context("paper")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```

```python
sns.set_style("ticks")
sns.set_context("poster")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```

```python
sns.set_style("ticks")
sns.set_context("talk")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```

```python
sns.set_style("ticks")
sns.set_context("notebook")
sns.scatterplot(data=penguins,x="flipper_length_mm",y="body_mass_g",hue="island")
sns.despine()
```
